### Day 4 – GROUP BY and HAVING Clause

#### Aggregate Functions

Aggregate functions are used to perform calculations on multiple rows of a table and return a single result. They help us find totals, averages, minimum values, maximum values and counts.

The commonly used aggregate functions are:
Function	Purpose
COUNT()	Counts the number of rows
SUM()	Calculates the total value
AVG()	Calculates the average value
MAX()	Finds the highest value
MIN()	Finds the lowest value

#### COUNT()

The COUNT() function is used to count the number of records in a table. It is useful when we want to know how many customers, products or orders are available.

```python
%%sql
SELECT COUNT(*) AS total_customers
FROM customers;
```

```python
%%sql
SELECT COUNT(*) AS total_products
FROM products;
```

### SUM()

The SUM() function is used to calculate the total of a numeric column. It is commonly used to calculate total sales, revenue or quantities.

```python
%%sql
SELECT SUM(order_amount) AS total_sales
FROM orders;
```

### AVG()

The AVG() function is used to calculate the average value of a numeric column. It helps us understand the average order amount or average product price.

```python
%%sql
SELECT AVG(order_amount) AS average_order_amount
FROM orders;
```

### MAX()

The MAX() function is used to find the highest value in a particular column. It helps us identify the highest order amount or most expensive product.

```python
%%sql
SELECT MAX(order_amount) AS highest_order
FROM orders;
```

### MIN()

The MIN() function is used to find the lowest value in a particular column. It helps us identify the lowest order amount or least expensive product.

```python
%%sql
SELECT MIN(unit_price) AS lowest_product_price
FROM products;
```

### GROUP BY Clause

The GROUP BY clause is used to group records that have the same values in a particular column. It is mainly used with aggregate functions to calculate summaries for each group.

For example, we can group orders based on their delivery partner, status or customers to understand the data category-wise.

```python
%%sql
SELECT city,
       COUNT(*) AS total_customers
FROM customers
GROUP BY city;
```

### GROUP BY with COUNT()
The COUNT() function is used with GROUP BY to count the number of records available in each group. It helps us understand how many records belong to a particular category.

Example: Count the number of products in each category.

```python
%%sql
SELECT category,
       COUNT(*) AS total_products
FROM products
GROUP BY category;
```

### Explanation:
- SELECT category – Displays the product category.
- COUNT(*) – Counts the number of products in each category.
- FROM products – Fetches records from the PRODUCTS table.
- GROUP BY category – Groups products having the same category.
Output understanding: This query displays each product category along with the total number of products available in that category.

### GROUP BY with SUM()
The SUM() function is used with GROUP BY to calculate the total value of a particular numeric column for each group. It is useful when we want to calculate total sales, revenue or quantity category-wise.

Example: Calculate total sales for each delivery partner.

```python
%%sql
SELECT partner,
       SUM(order_amount) AS total_sales
FROM orders
GROUP BY partner;
```

Explanation:
- SELECT partner – Displays the delivery partner name.
- SUM(order_amount) – Calculates the total order amount for each partner.
- FROM orders – Fetches records from the ORDERS table.
- GROUP BY partner – Groups orders handled by the same delivery partner.

This query displays each delivery partner along with the total sales amount generated through their orders.

What I understood: GROUP BY with SUM() helps us calculate totals separately for each group instead of calculating the grand total of the entire table.

### GROUP BY with AVG()
The AVG() function is used with GROUP BY to calculate the average value of a numeric column for each group. It helps us understand the average sales, order amount or product price category-wise.

Example: Calculate the average product price for each category.


```python
%%sql
SELECT category,
       AVG(unit_price) AS average_price
FROM products
GROUP BY category;
```

#### - SELECT category – Displays the product category.
- AVG(unit_price) – Calculates the average price of products in each category.
- FROM products – Fetches records from the PRODUCTS table.
- GROUP BY category – Groups products belonging to the same category.

This query displays each product category along with its average product price.

What I understood: GROUP BY with AVG() helps us compare the average values of different groups instead of calculating the average of the entire table.

### HAVING Clause
The HAVING clause is used to filter records after they have been grouped using the GROUP BY clause. It is mainly used with aggregate functions when we want to display only groups that satisfy a particular condition.

Example: Find delivery partners whose total order amount is greater than 50,000.

```python
%%sql
SELECT partner,
       SUM(order_amount) AS total_sales
FROM orders
GROUP BY partner
HAVING SUM(order_amount) > 50000;
```

#### Explanation:
- SELECT partner – Displays the delivery partner name.
- SUM(order_amount) – Calculates the total order amount for each partner.
- GROUP BY partner – Groups orders based on the delivery partner.
- HAVING SUM(order_amount) > 50000 – Displays only partners whose total order amount is greater than 50,000.

This query displays only those delivery partners whose total sales exceed 50,000.

GROUP BY first creates groups, and HAVING filters those groups based on the given condition.

### Difference Between WHERE and HAVING
WHERE is used to filter individual rows before grouping the data.

HAVING is used to filter groups after applying GROUP BY. It is commonly used with aggregate functions like SUM(), COUNT() and AVG().
