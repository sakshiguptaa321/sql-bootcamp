# Day 3 – Table Creation and SQL Commands

#### Table Creation

A table is used to store data in the form of rows and columns. The CREATE TABLE statement is used to create a new table by defining its column names and data types.

CREATE TABLE table_name (
    column_name1 datatype,
    column_name2 datatype
);

### Data Types in Oracle SQL 
A data type tells Oracle what kind of value can be stored in a column.
- NUMBER – Stores numeric values.
- CHAR – Stores fixed-length character data.
- VARCHAR2 – Stores variable-length character data.
- DATE – Stores date and time values.
- CLOB – Stores large character data, such as articles and documents.
- BLOB – Stores binary data, such as images, audio, video and PDFs.
- LONG – Stores large character data.

### DDL – Data Definition Language
Markdown cell

DDL stands for Data Definition Language. DDL commands are used to define and modify the structure of database objects such as tables. These commands help us create, change, rename and remove database tables.

The main DDL commands are CREATE, ALTER, DROP, RENAME and TRUNCATE.

### Display Records Using SELECT

The SELECT statement is used to retrieve and display records from an existing table. We can display all columns or select only the specific columns that we want to view.

```python
%%sql
SELECT * FROM customers;
```

### ALTER TABLE

The ALTER TABLE statement is used to modify the structure of an existing table. We can add new columns, modify existing columns or remove columns based on our requirements.

```python
%%sql
ALTER TABLE customers
ADD alternate_phone VARCHAR2(15);
```

### TRUNCATE TABLE

The TRUNCATE TABLE statement is used to remove all records from an existing table while keeping its structure intact. Unlike DELETE, it removes all rows at once and does not allow us to restore the deleted records using ROLLBACK.

Syntax is : TRUNCATE TABLE table_name;

### RENAME TABLE

The RENAME statement is used to change the name of an existing table in the database. It changes only the table name, while the table structure and stored records remain the same.

Syntax is : RENAME old_table_name TO new_table_name;

### DROP TABLE

The DROP TABLE statement is used to completely remove an existing table from the database. It deletes both the table structure and all the records stored inside it.

Syntax is : 
DROP TABLE table_name;
