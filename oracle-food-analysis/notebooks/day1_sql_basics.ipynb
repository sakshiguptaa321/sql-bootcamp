# Day 1 – SQL Basics

Today, I learned the basics of SQL (Structured Query Language) and how it is used to communicate with relational databases.

SQL helps us create tables, insert records, update data, delete records and retrieve information from a database.

### What is Data?
Data is a collection of raw facts and figures.

### What is a Database?
A database is an organized collection of related data.

### What is DBMS?
DBMS stands for Database Management System. It is software used to create, store, manage, retrieve and update data in a database.

Examples: Oracle, MySQL, SQL Server and PostgreSQL.

### What is RDBMS?
RDBMS stands for Relational Database Management System. It stores data in tables consisting of rows and columns. Tables can be connected using Primary Keys and Foreign Keys.

### Primary Key
A Primary Key uniquely identifies each record in a table.

### Foreign Key
A Foreign Key is used to establish a relationship between two tables by referencing the Primary Key of another table.

#### Understanding our database tables

In my food-analysis project I am working with four tables:
Table	Purpose

PRODUCTS	Stores product details
ORDERS	Stores order information
ORDER_ITEMS	Stores individual order items

These tables are connected using Primary Key and Foreign Key relationships.

## 1. SELECT Statement
The SELECT statement is used to retrieve data from a database.

The * symbol is used to display all columns from a table.

```python
%%sql
SELECT * FROM customers FETCH FIRST 5 ROWS ONLY;
```

### FROM Clause
The FROM clause specifies the table from which we want to retrieve data.

```python
%%sql
SELECT customer_name
FROM customers
FETCH FIRST 5 ROWS ONLY;
```

### WHERE Clause
The WHERE clause is used to filter records based on a particular condition.

It helps us retrieve only the records that match our requirements.

```python
%%sql
SELECT *
FROM customers
WHERE city = 'Bengaluru'
FETCH FIRST 5 ROWS ONLY;
```

### Special Operators in SQL
I learned different operators that help filter records based on specific conditions.

A. Equal To (=)

Used to retrieve records that match a particular value.

```python
%%sql
SELECT *
FROM customers
WHERE customer_id = 'C0001';
```

### This query retrieves the customer whose ID is C0001.

### Not Equal To ( <> )

Used to retrieve records that do not match a particular value.

```python
%%sql
SELECT *
FROM products
WHERE category <> 'Beverages';
```

#### This query displays products that do not belong to the Beverages category.

### IN Operator

The IN operator checks whether a value matches any value in a given list.

```python
%%sql
SELECT *
FROM customers
WHERE city IN ('Bengaluru', 'Mysuru')
FETCH FIRST 5 ROWS ONLY;
```

### NOT IN Operator

The NOT IN operator excludes the specified values from the result.

```python
%%sql
SELECT *
FROM products
WHERE category NOT IN ('Momos', 'Starters');
```

### BETWEEN Operator

The BETWEEN operator is used to retrieve values within a specified range. Both starting and ending values are included.

```python
%%sql
SELECT *
FROM products
WHERE unit_price BETWEEN 100 AND 150;
```

### NOT BETWEEN Operator

The NOT BETWEEN operator retrieves values outside the specified range.

```python
%%sql
SELECT *
FROM products
WHERE unit_price NOT BETWEEN 130 AND 180;
```

### LIKE Operator

The LIKE operator is used for pattern matching in text values.
- % represents zero or more characters.
- _ represents exactly one character.

```python
%%sql
SELECT *
FROM customers
WHERE customer_name LIKE 'S%';
```

### NOT LIKE Operator

The NOT LIKE operator retrieves records that do not match a specified pattern.

```python
%%sql
SELECT *
FROM customers
WHERE customer_name NOT LIKE 'S%';
```

### AND Operator

The AND operator is used when all the given conditions must be TRUE.

```python
%%sql
SELECT *
FROM products
WHERE category = 'Momos'
AND unit_price > 150;
```

### OR Operator
The OR operator is used to filter records when at least one of the given conditions must be TRUE. If either condition matches, the record will be included in the result.

```python
%%sql
SELECT *
FROM products
WHERE category = 'Starters'
OR unit_price = 210;
```

### UPPER() Function
The UPPER() function is used to convert all lowercase and mixed-case letters in a string into uppercase letters. It helps display text in capital letters without changing the original data stored in the table.

```python
%%sql
SELECT
    customer_name,
    UPPER(customer_name) AS upper_name
FROM customers
FETCH FIRST 5 ROWS ONLY;
```

### LOWER() Function
The LOWER() function is used to convert all uppercase and mixed-case letters in a string into lowercase letters. It is useful when we want to display text in small letters.

```python
%%sql
SELECT
customer_name,
LOWER(customer_name) AS lower_name
FROM customers
FETCH FIRST 5 ROWS ONLY;
```
