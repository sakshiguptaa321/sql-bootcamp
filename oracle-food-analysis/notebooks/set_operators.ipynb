### Set Operators In SQL

### Set Operators are - 
Set operators are used to combine the results of two or more SELECT queries

Oracle SQL provides four main operators 
1. UNION
2. UNION ALL
3. INTERSECT
4. MINUS
Some important rules are -
- Both SELECT queries must return the same number of columns
- Corresponding columns should have compatible data types

### UNION
 UNION combines the results of two queries and removes duplicate records.

```python
%%sql
SELECT customer_id
FROM customers
UNION
SELECT customer_id
FROM orders;
```

### Understanding of above query - 
The first query gets customer IDs from the customers table

The second query gets customer IDs from the orders table

UNION combines both results

Duplicate customer IDs are removed

### Union ALL - UNION ALL combines the results of two queries and keeps duplicate records

```python
%%sql
SELECT customer_id
FROM customers

UNION ALL

SELECT customer_id
FROM orders;
```

### Understanding -
Combines customer IDs from both tables

Duplicate values are also displayed

Unlike UNION UNION ALL does not remove duplicates

### INTERSECT
 INTERSECT returns only the common records that are present in both query results

```python
%%sql
SELECT customer_id
FROM customers

INTERSECT

SELECT customer_id
FROM orders;
```

### Understanding - 
Gets customer IDs from both tables

Displays only IDs that are present in both results

Duplicate records are removed

### MINUS
MINUS returns records from the first query that are not present in the second query

```python
%%sql
SELECT customer_id
FROM customers

MINUS

SELECT customer_id
FROM orders;
```

### Understanding
- First query gets all customer IDs
- Second query gets customer IDs from orders
- MINUS displays customer IDs that are present in customers but not in orders
