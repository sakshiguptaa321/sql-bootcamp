### JOINS IN SQL

### Introduction to JOINS
A JOIN in SQL is used to combine data from two or more tables based on a related column between them.

In our database, the customers and orders tables are connected through the common column customer_id.

### INNER JOIN
 INNER JOIN returns only the rows that have matching values in both tables.

```python
%%sql
SELECT customer_name, order_id
FROM customers
INNER JOIN orders
ON customers.customer_id = orders.customer_id;
```

### Explanation
- customers and orders are the two tables being joined.
- ON specifies the matching condition.
- Only matching customer and order records are displayed.

### LEFT OUTER JOIN
LEFT JOIN returns all rows from the left table and matching rows from the right table. If there is no match, the right table's columns show NULL.

```python
%%sql
SELECT customer_name, order_id
FROM customers
LEFT JOIN orders
ON customers.customer_id = orders.customer_id;
```

### Explanation:
- All customers are displayed.
- If a customer has placed an order, the order ID is displayed.
- If there is no matching order, the order ID shows NULL.


### RIGHT OUTER JOIN
RIGHT JOIN returns all rows from the right table and matching rows from the left table. If there is no match, the left table's columns show NULL.

```python
%%sql
SELECT customer_name, order_id
FROM customers
RIGHT JOIN orders
ON customers.customer_id = orders.customer_id;
```

### Explanation:
- All orders are displayed.
- Matching customer names are displayed.
- If an order has no matching customer, the customer name shows NULL.

### FULL OUTER JOIN
 FULL OUTER JOIN returns all rows from both tables. It includes matching and non-matching records.

```python
%%sql
SELECT customer_name, order_id
FROM customers
FULL OUTER JOIN orders
ON customers.customer_id = orders.customer_id;
```

### Explanation:
- Displays all customers and all orders.
- Matching records are combined.
- Non-matching records show NULL for the missing side.

### SELF JOIN
Definition: SELF JOIN means joining a table with itself. It is used when we need to compare or connect rows within the same table.

Example: SELF JOIN is commonly used to connect employees with their managers.

### CROSS JOIN
CROSS JOIN returns all possible combinations of rows from two tables. Each row of the first table is combined with every row of the second table.



```python
%%sql
SELECT customer_name, product_name
FROM customers
CROSS JOIN products;
```

### JOIN Multiple Tables
Joining multiple tables means combining data from more than two tables in a single SQL query.
