# Reporting Views

This notebook creates Oracle reporting views for the `FOOD_ANALYSIS` schema.

Views included:
- `VW_TOP_10_CUSTOMERS_BY_SALES`: top 10 customers by total order sales.
- `VW_TOP_10_CITIES_BY_SALES`: top 10 cities by total order sales.

```python
%load_ext sql
import os
from dotenv import load_dotenv
from sqlalchemy.engine import URL

load_dotenv()
connection_url = URL.create(
    "oracle+oracledb",
    username=os.environ["ORACLE_USER"],
    password=os.environ["ORACLE_PASSWORD"],
    host="localhost",
    port=1522,
    query={"service_name": "FREEPDB1"},
).render_as_string(hide_password=False)
%sql $connection_url
```

## 1. Top 10 customers by sales

This view calculates total sales per customer using `SUM(o.order_amount)` from `ORDERS`, joins customer details from `CUSTOMERS`, and returns the 10 customers with the highest total sales.

```python
%%sql
CREATE OR REPLACE VIEW vw_top_10_customers_by_sales AS
SELECT
    c.customer_id,
    c.customer_name,
    c.city,
    COUNT(o.order_id) AS total_orders,
    SUM(o.order_amount) AS total_sales
FROM customers c
JOIN orders o
    ON c.customer_id = o.customer_id
GROUP BY
    c.customer_id,
    c.customer_name,
    c.city
ORDER BY total_sales DESC
FETCH FIRST 10 ROWS ONLY
```

```python
%%sql
SELECT *
FROM vw_top_10_customers_by_sales
ORDER BY total_sales DESC
```

## 2. Top 10 cities by total sales

This view groups orders by customer city and calculates total sales for each city. It returns the 10 cities with the highest total sales.

```python
%%sql
CREATE OR REPLACE VIEW vw_top_10_cities_by_sales AS
SELECT
    c.city,
    COUNT(DISTINCT c.customer_id) AS total_customers,
    COUNT(o.order_id) AS total_orders,
    SUM(o.order_amount) AS total_sales
FROM customers c
JOIN orders o
    ON c.customer_id = o.customer_id
GROUP BY c.city
ORDER BY total_sales DESC
FETCH FIRST 10 ROWS ONLY
```

```python
%%sql
SELECT *
FROM vw_top_10_cities_by_sales
ORDER BY total_sales DESC
```
