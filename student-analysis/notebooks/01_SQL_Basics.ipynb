## What is RDBMS?

RDBMS stands for Relational Database Management System.

It stores data in tables consisting of rows and columns.

It also establishes relationships between tables using keys.

### Primary Key

A Primary Key is a column that uniquely identifies
each record in a table. It should be unique and cannot be null.

Example:
In a Students table
  Student_ID can be the Primary Key as all students have different ID's

### Foreign Key

A Foreign Key is a column that refers to the
Primary Key of another table.

It establishes relationships between two tables.

## 5. Basic SQL Clauses

A SQL query uses different clauses to retrieve
and filter data.

The three basic clauses are:

1. SELECT is used to retrieve the data.
2. FROM is used to fetch data where from.
3. WHERE is used to filter rows based on a given condition.

### A. SELECT Clause

SELECT is used to choose the columns we want
to display from a table.

Syntax:

SELECT column_name
FROM table_name;

Example:

SELECT first_name, salary
FROM employees;

To select all columns:

SELECT *
FROM employees;

Note:
* represents all columns in the table.

### B. FROM Clause

FROM specifies the table from which we want
to retrieve data.

Syntax:

SELECT *
FROM table_name;

Example:

SELECT *
FROM employees;

### C. WHERE Clause

WHERE is used to filter records based on a condition.

Syntax:

SELECT column_name
FROM table_name
WHERE condition;

Example:

SELECT *
FROM employees
WHERE salary > 10000;

Explanation:

This query retrieves only those employees
whose salary is greater than 10000.
