# SQL Dunctions 

##### ORDER BY - ORDER BY is used to sort the result of a SELECT query in either ascending (ASC) or descending (DESC) order.

```python
%%sql
SELECT *
FROM Students
ORDER BY Age DESC;
```

```python
%%sql
SELECT * FROM faculty
ORDER BY salary ASC;
```

# Case Manipulation Functions

UPPER() - It is used to convert all text to uppercase 
LOWER() - It is used to convert all tect to lowercase

```python
%%sql
SELECT first_name,
       UPPER(first_name) AS Upper_Name,
       LOWER(first_name) AS Lower_Name
FROM Students;
```
