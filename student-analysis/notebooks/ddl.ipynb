# CREATE Statements
CREATE statements help to create new table in the databases.

### Departments Table

```python
%%sql
CREATE TABLE departments (
    dept_id integer primary key,
    dept_name text
)
```

### Courses Table

```python
%%sql
CREATE TABLE courses (
    course_id integer primary key,
    course_name text,
    credits integer,
    dept_id integer,
    FOREIGN KEY (dept_id) REFERENCES departments(dept_id)
)
```

### Students Table

```python
%%sql
CREATE TABLE students (
    student_id integer primary key,
    first_name text,
    last_name text,
    age integer,
    gender text,
    email text,
    city text,
    dept_id integer,
    FOREIGN KEY (dept_id) REFERENCES departments(dept_id)
)
```

# ALTER Statements
ALTER statements help with altering table structure for adding a new column, modifying column or drop column. 

```python
%%sql
ALTER TABLE students
ADD COLUMN admission_date datetime;
```

```python
%%sql
CREATE TABLE student_marks (
    marks_id integer primary key,
    student_id integer,
    course_id integer,
    marks_obtained integer,
    total_marks integer,
    exam_date datetime,
    FOREIGN KEY (student_id) REFERENCES students(student_id),
    FOREIGN KEY (course_id) REFERENCES courses(course_id)
)
```

# Drop Statements
Drop command drop the data and table along with it but TRUNCATE only drop the data not the table itself. 

```python
%%sql
DROP TABLE IF EXISTS departments;
DROP TABLE IF EXISTS courses;
DROP TABLE IF EXISTS students;
```
