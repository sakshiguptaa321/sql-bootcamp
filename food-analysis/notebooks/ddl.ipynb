# DDL: Create the schema

**DDL** means **Data Definition Language**. It is the part of SQL used to define or change database structures such as tables and constraints. This notebook uses `CREATE TABLE` and `DROP TABLE`; the next notebook uses DML (`INSERT`) to add rows.

Run this notebook first. It creates the SQLite tables and resets any existing exercise data.

```python
%load_ext sql
%sql sqlite:///food-analysis/food_analysis.db
```

## DDL statements and concepts

- `PRAGMA foreign_keys = ON` is a SQLite setting that makes this connection enforce foreign-key rules.
- `DROP TABLE IF EXISTS table_name` removes a table and all its rows. `IF EXISTS` avoids an error if the table is not there yet. Child tables are dropped before their parent tables.
- `CREATE TABLE table_name (...)` creates a table. Inside the parentheses, each definition gives a column name, its type, and optional constraints.
- `TEXT` stores text, `INTEGER` stores whole numbers, and `REAL` stores numeric values that may have decimals. SQLite uses flexible type affinity; these types describe the intended data.
- `PRIMARY KEY` identifies each row uniquely. Here, IDs such as `customer_id` are the primary keys.
- `NOT NULL` requires a value for that column.
- `REFERENCES parent_table(id)` creates a foreign key: a value must match an existing key in the parent table when foreign-key enforcement is enabled.

This DDL drops and recreates the exercise tables, so running it again resets their data.

## What each table represents

- `customers` stores one row per customer. `customer_id` uniquely identifies a customer; name and ID are required.
- `products` stores one row per product, including its category and listed price.
- `orders` stores one row per order. `customer_id` links each order to the customer who placed it.
- `order_items` stores the products and quantities within each order. Its `order_id` and `product_id` link to the order and product, so one order can contain multiple products.

These links are **relationships**: orders depend on customers, and order items depend on both orders and products. That is why the parent tables are created before their child tables.

```python
%%sql
-- Schema setup (DDL). Running this script resets the exercise tables.
PRAGMA foreign_keys = ON;

DROP TABLE IF EXISTS order_items;
DROP TABLE IF EXISTS orders;
DROP TABLE IF EXISTS products;
DROP TABLE IF EXISTS customers;
DROP TABLE IF EXISTS lessons;

CREATE TABLE customers (
    customer_id TEXT PRIMARY KEY,
    customer_name TEXT NOT NULL,
    city TEXT,
    area TEXT,
    signup_date TEXT
);

CREATE TABLE products (
    product_id TEXT PRIMARY KEY,
    product_name TEXT NOT NULL,
    category TEXT,
    unit_price REAL
);

CREATE TABLE orders (
    order_id TEXT PRIMARY KEY,
    order_placed_at TEXT,
    status TEXT,
    partner TEXT,
    order_amount REAL,
    customer_id TEXT NOT NULL REFERENCES customers(customer_id)
);

CREATE TABLE order_items (
    order_item_id TEXT PRIMARY KEY,
    order_id TEXT NOT NULL REFERENCES orders(order_id),
    product_id TEXT NOT NULL REFERENCES products(product_id),
    quantity INTEGER NOT NULL,
    unit_price REAL NOT NULL,
    line_amount REAL NOT NULL
);
```
