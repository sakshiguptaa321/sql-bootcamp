## Concepts

- Aggregate functions calculate a value across rows: `COUNT` counts rows, `SUM` adds values, `AVG` calculates an average, and `MIN`/`MAX` find the smallest/largest value.
- `GROUP BY` puts rows with the same value into groups so aggregates can be calculated for each group. Every selected column that is not aggregated should be included in `GROUP BY`.
- `HAVING` filters groups after aggregation. `WHERE` filters individual rows before grouping.
- `ROUND(value, 2)` displays a numeric result rounded to two decimal places.

# Raw Orders
This is just raw orders for preview purpose only. 

```python
%%sql
select *from orders 
limit 10
```

## Count orders and calculate totals for each status

```python
%%sql
SELECT status,
       COUNT(*) AS order_count,
       ROUND(SUM(order_amount), 2) AS total_order_amount,
       ROUND(AVG(order_amount), 2) AS average_order_amount
FROM orders
GROUP BY status
ORDER BY order_count DESC;
```

## Keep only statuses with at least 50 orders

`HAVING` applies the condition to each completed group.

```python
%%sql
SELECT status, COUNT(*) AS order_count
FROM orders
GROUP BY status
HAVING COUNT(*) >= 50
ORDER BY order_count DESC;
```

# Partner Month Summary
This report is for understanding how many orders each partner source month-on-month basics. 

```python
%%sql
select strftime('%Y-%m',order_placed_at) as order_month, partner, count(partner) as partner_count
from orders
group by partner, strftime('%Y-%m',order_placed_at)
order by order_month desc
```
