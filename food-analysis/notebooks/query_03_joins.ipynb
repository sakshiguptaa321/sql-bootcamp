## Concepts

- `JOIN` (or `INNER JOIN`) returns rows where the join condition matches in both tables.
- `ON` states how rows match, usually by connecting a foreign key to a primary key.
- Table aliases such as `o` and `c` are short names that make queries easier to read. For example, `o.order_id` refers to the `order_id` column from `orders`.
- Joining `orders` to `customers` adds customer details to each order. Joining `order_items` to `products` adds product names to order-line data.

## Show delivered orders with customer details

```python
%%sql
SELECT o.order_id, c.customer_name, c.area,
       o.order_placed_at, o.order_amount
FROM orders AS o
JOIN customers AS c ON c.customer_id = o.customer_id
WHERE o.status = 'delivered'
ORDER BY o.order_placed_at DESC
LIMIT 10;
```

## Rank products by units sold in delivered orders

This query joins three tables, then groups the matching order-item rows by product.

```python
%%sql
SELECT p.product_name,
       SUM(oi.quantity) AS units_sold,
       ROUND(SUM(oi.line_amount), 2) AS item_sales
FROM order_items AS oi
JOIN products AS p ON p.product_id = oi.product_id
JOIN orders AS o ON o.order_id = oi.order_id
WHERE o.status = 'delivered'
GROUP BY p.product_id, p.product_name
ORDER BY units_sold DESC
LIMIT 10;
```
