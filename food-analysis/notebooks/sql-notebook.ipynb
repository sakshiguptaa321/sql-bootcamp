# Food orders SQL practice

Use the notebooks in this order. They all connect to the same local `food_analysis.db` SQLite database:

1. [ddl.ipynb](./ddl.ipynb) — create the schema (resets the exercise tables).
2. [dml.ipynb](./dml.ipynb) — insert the practice data with SQL statements.
3. [queries.ipynb](./queries.ipynb) — explore the data with SQL queries.

Select the **Python (learning-sql)** kernel in each notebook. Run the DDL notebook before DML, and rerun DDL before reseeding.
