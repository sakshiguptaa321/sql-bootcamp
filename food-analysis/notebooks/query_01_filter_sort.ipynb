## Concepts

- `SELECT` chooses which columns to show; `FROM` names the table to read.
- `DISTINCT` removes duplicate values from the result.
- `WHERE` filters rows before they are returned. Use comparison operators such as `=`, `>`, and `>=`; combine conditions with `AND` or `OR`.
- `ORDER BY` sorts the result. `ASC` means ascending (the default); `DESC` means descending.
- `LIMIT` caps how many rows are returned.

## List the available order statuses

```python
%%sql
SELECT DISTINCT status
FROM orders
ORDER BY status ASC;
```

## Find recent high-value delivered orders

The `WHERE` clause requires both conditions to be true. The most expensive matching orders appear first.

```python
%%sql
SELECT order_id, order_placed_at, order_amount
FROM orders
WHERE status = 'delivered' AND order_amount >= 500
ORDER BY order_amount DESC
LIMIT 10;
```
